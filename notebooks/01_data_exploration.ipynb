```python
from pathlib import Path

ROOT = Path("/content/drive/MyDrive/synthetic-mri-motion-detection")

RAW_DIR = ROOT / "data" / "raw"
PROCESSED_DIR = ROOT / "data" / "processed"
SYNTHETIC_DIR = ROOT / "data" / "synthetic"
MODEL_DIR = ROOT / "outputs" / "models"
METRICS_DIR = ROOT / "outputs" / "metrics"
FIGURE_DIR = ROOT / "outputs" / "figures"

ARCHIVE_PATH = RAW_DIR / "IXI-T1.tar"
SAMPLE_DIR = RAW_DIR / "sample"

SAMPLE_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", ROOT)
print("Archive path:", ARCHIVE_PATH)
print("Sample directory:", SAMPLE_DIR)
```

```python
import urllib.request
from tqdm.auto import tqdm

IXI_URL = "http://biomedic.doc.ic.ac.uk/brain-development/downloads/IXI/IXI-T1.tar"

if ARCHIVE_PATH.exists():
    print("Archive already downloaded.")
else:
    print("Downloading IXI-T1 archive. This may take several minutes.")

    def progress(block_num, block_size, total_size):
        downloaded = block_num * block_size
        if total_size > 0:
            percent = min(100, downloaded * 100 / total_size)
            print(f"\rDownloaded: {percent:.1f}%", end="")

    urllib.request.urlretrieve(IXI_URL, ARCHIVE_PATH, reporthook=progress)

print("\nArchive size (GB):", round(ARCHIVE_PATH.stat().st_size / 1e9, 2))
```

```python
import tarfile
import os

MAX_FILES = 10
extracted = 0

try:
    # Opening with ignore_zeros=True can sometimes help bypass minor stream corruption
    with tarfile.open(ARCHIVE_PATH, "r", ignore_zeros=True) as tar:
        print("Reading archive members...")
        for member in tar:
            if member.isfile() and member.name.endswith((".nii", ".nii.gz")):
                tar.extract(member, path=SAMPLE_DIR)
                extracted += 1

                if extracted >= MAX_FILES:
                    break
except (tarfile.ReadError, EOFError) as e:
    print(f"\nError reading tar file: {e}")
    print("\nThis typically means the downloaded archive is incomplete or corrupt.")
    print(f"Current archive size is only {round(ARCHIVE_PATH.stat().st_size / 1e9, 2)} GB.")
    print("Please delete the corrupted file and re-run the download cell:")
    print(f"!rm {ARCHIVE_PATH}")

print(f"\nExtracted {extracted} NIfTI files.")
print("Sample directory contents:")
for file in sorted(SAMPLE_DIR.rglob("*")):
    if file.is_file():
        print("-", file.name)
```

```python
from pathlib import Path
import glob

ROOT = Path("/content/drive/MyDrive/synthetic-mri-motion-detection")

RAW_SAMPLE_DIR = ROOT / "data" / "raw" / "sample"
PROCESSED_DIR = ROOT / "data" / "processed"
FIGURE_DIR = ROOT / "outputs" / "figures"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

nii_files = sorted(
    glob.glob(str(RAW_SAMPLE_DIR / "*.nii"))
    + glob.glob(str(RAW_SAMPLE_DIR / "*.nii.gz"))
)

print("Found MRI volumes:", len(nii_files))
for file in nii_files:
    print("-", Path(file).name)
```

```python
import numpy as np

IMAGE_SIZE = 224
SLICE_AXIS = 2  # axial slices
LOW_PERCENTILE = 1
HIGH_PERCENTILE = 99
MIN_BRAIN_PIXELS = 500

print("Target image size:", IMAGE_SIZE)
print("Slice axis:", SLICE_AXIS)
```

```python
import cv2
import nibabel as nib

def preprocess_volume(nifti_path):
    img = nib.load(nifti_path)
    volume = img.get_fdata().astype(np.float32)

    slices = []
    metadata = []

    for slice_index in range(volume.shape[SLICE_AXIS]):
        if SLICE_AXIS == 2:
            slice_img = volume[:, :, slice_index]
        elif SLICE_AXIS == 1:
            slice_img = volume[:, slice_index, :]
        else:
            slice_img = volume[slice_index, :, :]

        # Remove mostly empty slices
        if np.count_nonzero(slice_img > 10) < MIN_BRAIN_PIXELS:
            continue

        # Percentile clipping
        low = np.percentile(slice_img, LOW_PERCENTILE)
        high = np.percentile(slice_img, HIGH_PERCENTILE)

        if high <= low:
            continue

        slice_img = np.clip(slice_img, low, high)
        slice_img = (slice_img - low) / (high - low)

        # Resize to model input size
        slice_img = cv2.resize(
            slice_img,
            (IMAGE_SIZE, IMAGE_SIZE),
            interpolation=cv2.INTER_LINEAR
        )

        slices.append(slice_img.astype(np.float32))
        metadata.append({
            "source_file": Path(nifti_path).name,
            "slice_index": slice_index,
            "label": "clean",
            "severity": "none"
        })

    return slices, metadata
```

```python
import pandas as pd

all_slices = []
all_metadata = []

for nifti_path in nii_files:
    slices, metadata = preprocess_volume(nifti_path)

    subject_id = Path(nifti_path).stem.replace("-T1", "")

    for local_index, (slice_img, meta) in enumerate(zip(slices, metadata)):
        output_path = PROCESSED_DIR / f"{subject_id}_axial_{local_index:03d}.npy"
        np.save(output_path, slice_img)

        meta["subject_id"] = subject_id
        meta["processed_path"] = str(output_path)
        all_metadata.append(meta)

    print(f"{subject_id}: {len(slices)} usable axial slices")

metadata_df = pd.DataFrame(all_metadata)
metadata_df.to_csv(PROCESSED_DIR / "slice_metadata.csv", index=False)

print("\nTotal processed slices:", len(metadata_df))
print("Metadata saved to:", PROCESSED_DIR / "slice_metadata.csv")
metadata_df.head()
```

```python
import cv2
import nibabel as nib
import numpy as np
from nibabel import as_closest_canonical

def preprocess_volume(nifti_path):
    img = nib.load(nifti_path)
    img = as_closest_canonical(img)
    volume = img.get_fdata().astype(np.float32)

    slices = []
    metadata = []

    # Canonical orientation: axial slices along the final axis
    for slice_index in range(volume.shape[2]):
        slice_img = volume[:, :, slice_index]

        # Ignore mostly empty slices
        if np.count_nonzero(slice_img > 10) < MIN_BRAIN_PIXELS:
            continue

        low = np.percentile(slice_img, LOW_PERCENTILE)
        high = np.percentile(slice_img, HIGH_PERCENTILE)

        if high <= low:
            continue

        slice_img = np.clip(slice_img, low, high)
        slice_img = (slice_img - low) / (high - low)

        slice_img = cv2.resize(
            slice_img,
            (IMAGE_SIZE, IMAGE_SIZE),
            interpolation=cv2.INTER_LINEAR
        )

        slices.append(slice_img.astype(np.float32))
        metadata.append({
            "source_file": Path(nifti_path).name,
            "slice_index": slice_index,
            "label": "clean",
            "severity": "none",
        })

    return slices, metadata
```

```python
import pandas as pd

selected_parts = []

for subject_id, subject_df in metadata_df.groupby("subject_id"):
    subject_df = subject_df.sort_values("slice_index").reset_index(drop=True)

    start = int(len(subject_df) * 0.20)
    end = int(len(subject_df) * 0.80)

    selected_subject_df = subject_df.iloc[start:end].copy()
    selected_parts.append(selected_subject_df)

central_metadata_df = pd.concat(selected_parts, ignore_index=True)

central_metadata_df.to_csv(
    PROCESSED_DIR / "central_slice_metadata.csv",
    index=False
)

print("Total retained central slices:", len(central_metadata_df))
print("Number of subjects:", central_metadata_df["subject_id"].nunique())
print("\nSelected slices per subject:")
print(central_metadata_df["subject_id"].value_counts().sort_index())

print("\nRange per subject:")
print(
    central_metadata_df.groupby("subject_id")["slice_index"]
    .agg(["min", "max", "count"])
    .sort_index()
)
```
